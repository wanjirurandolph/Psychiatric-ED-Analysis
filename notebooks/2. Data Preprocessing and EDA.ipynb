# Psychiatric ER Length of Stay: 
Which Hospitals Have the Longest Mental Health ED Stays ?

**Data:** CMS Hospital Compare — Timely and Effective Care + Hospital General Information


## What is being measured
`wait_psych` is CMS measure OP-18c: the median time from ED arrival to ED departure
for psychiatric/mental health patients who are discharged from the ED. 
Patients admitted to the hospital are not included.
`wait_regular` is OP-18b: the same measure for ED patients, excluding
psychiatric/mental health patients and patients transferred to another facility.
These measure ED length of stay, not boarding time.



## Research Question
Which hospital characteristics predict the longest ED waits for mental health patients?


## Why It Matters
When mental health patients wait Long in the ED
it delays their care and crowds the ED for everyone else.

1. Load the dataset built in SQL (`1. SQL_Dataset.ipynb`)
2. Inspect and clean the data
3. Explore the data
4. Feature engineering

## Feature Engineering
This section prepares the data for modeling: creating the target, grouping
categories that are too small or too detailed, and filling blanks that have
a meaning. Blanks in the numeric features are filled later, after the
train/test split, so information from the test set doesn't leak into training.

### Target: severe ED stay
The goal is to identify hospitals with the longest mental health ED stays,
so the target is binary: `severe` = 1 if a hospital's `wait_psych` is in the
top 25% (at or above the 75th percentile), and 0 otherwise.

Why a top-25% cutoff:
- It matches the question: the focus is on the hospitals where stays are longest,
  not on predicting the exact number of minutes.
- Quartiles are a common way to benchmark performance, so "top 25%" is easy to explain.
- A binary target is not affected by the extreme outliers (up to 6,870 minutes).

This cutoff is a modeling choice, not a clinical standard for this exact measure

```python
#cutoff for the top 25% of mental health ED stays (75th percentile)
cutoff = df["wait_psych"].quantile(0.75)
print(cutoff)
```

```python
#severe = 1 if the hospital is in the top 25%, otherwise 0
#astype(int) turns True/False into 1/0
df["severe"] = (df["wait_psych"] >= cutoff).astype(int)

#check the counts (should be about 25% / 75%)
print(df["severe"].value_counts())
print(df["severe"].value_counts(normalize=True))
```

The cutoff is 324 minutes. 759 hospitals (25%) are labeled severe and 2,259 (75%) are not.
The classes are imbalanced (about 1 severe for every 3 not severe), so accuracy alone
would be misleading: a model that always guesses "not severe" would be 75% accurate
while finding none of the severe hospitals. This is why the models will be judged
on recall and AUC, compared against a baseline that always guesses "not severe."

### Region
`State` has 51 values (50 states + DC). Using it directly would create 51 separate
columns in the model, many with only a few hospitals, which makes the results
unstable. Instead, states are grouped into the 4 U.S. Census Bureau regions
(Northeast, Midwest, South, West). This keeps the geographic information
while giving each group enough hospitals to compare.

```python
#U.S. Census Bureau regions
northeast = ["CT", "ME", "MA", "NH", "RI", "VT", "NJ", "NY", "PA"]
midwest = ["IL", "IN", "MI", "OH", "WI", "IA", "KS", "MN", "MO", "NE", "ND", "SD"]
south = ["DE", "DC", "FL", "GA", "MD", "NC", "SC", "VA", "WV",
         "AL", "KY", "MS", "TN", "AR", "LA", "OK", "TX"]
west = ["AZ", "CO", "ID", "MT", "NV", "NM", "UT", "WY",
        "AK", "CA", "HI", "OR", "WA"]

#make a lookup: each state -> its region
region_map = {}
for s in northeast: region_map[s] = "Northeast"
for s in midwest: region_map[s] = "Midwest"
for s in south: region_map[s] = "South"
for s in west: region_map[s] = "West"

#create the region column
df["region"] = df["State"].map(region_map)
```

```python
#every hospital should have a region (0 = none missed)
print(df["region"].isna().sum())

#how many hospitals in each region
df["region"].value_counts()
```

All 3,018 hospitals were assigned a region. The Midwest (1,007) and South (972) have
the most hospitals, followed by the West (630) and Northeast (409). Every region has
enough hospitals to compare.

### Ownership group
`Hospital Ownership` has 10 types, and some are very small (Tribal 1, Physician 10,
Federal 12). They are grouped into 3 broader types so every group has enough hospitals:
- **Non-profit:** voluntary non-profit (private, church, other)
- **Government:** hospital district, local, state, federal, and tribal
- **For-profit:** proprietary and physician-owned

```python
#group ownership into 3 broader types
def group_ownership(x):
    if x.startswith("Voluntary non-profit"):
        return "Non-profit"
    elif x.startswith("Government") or x == "Tribal":
        return "Government"
    elif x in ["Proprietary", "Physician"]:
        return "For-profit"
    else:
        return None   #anything unexpected shows up as blank so we can catch it

df["ownership_group"] = df["Hospital Ownership"].apply(group_ownership)
```

```python
#every hospital should have a group (0 = none missed)
print(df["ownership_group"].isna().sum())
df["ownership_group"].value_counts()
```

Most hospitals are non-profit (1,999), followed by government (592) and for-profit (427).
All three groups are large enough for the model to compare.
